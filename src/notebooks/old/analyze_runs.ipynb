# ICL experiment analysis

Cross-run view over `runs/*` produced by `ExperimentTracker`. Best-so-far curves, success
rates, the experiment index, and per-run drill-down.

Run from the `src/` directory with the analysis extra installed:
`pip install -e ".[analysis]"`. Set `RUNS_DIR` below to your runs folder.

```python
%matplotlib inline
import matplotlib.pyplot as plt
from results.analysis import (
    load_index, load_progress, load_solutions, load_avg_valid_value,
    plot_best_so_far, plot_valid_percentage, plot_avg_valid_value, plot_success_rate,
)

RUNS_DIR = "../runs"   # change to an absolute path if running the notebook elsewhere
```

## Experiment index
One row per run: problem, knobs, best result, status.

```python
idx = load_index(RUNS_DIR)
idx   # the `run` column holds the directory names you use to select runs below
```

## Best-so-far by generation
One line per selected run. Minimize problems (erdos, ac1) trend **down** (lower native metric is better). Saves a PNG via `save_path`.

```python
RUNS ={"cp_26_no_icl_g5x12_gen30v2": "No ICL", "cp_26_random_n_30_g5x12_gen30v2": "Random", "cp_26_contrastive_n_30_g5x12_gen30v2": "Best and Worst"}
plot_best_so_far(RUNS_DIR, runs=RUNS, save_path="best_so_far.png")
plt.show()
```

## Valid solutions (%) by generation
Percentage of the generated candidates that were valid, per generation, for each selected run.

```python
plot_valid_percentage(RUNS_DIR, runs=RUNS, save_path="valid_percentage.png")
plt.show()
```

## Average value of valid solutions by generation
Mean objective `value` over each generation's valid solutions, read from each run's
`solutions/manifest.jsonl`. Shows the *typical* quality of proposals per generation
(vs. best-so-far, which tracks only the frontier).

```python
plot_avg_valid_value(RUNS_DIR, runs=RUNS, save_path="avg_valid_value.png")
plt.show()
```

## Success rate by generation
Fraction of the group_size candidates that were valid, per generation.

```python
plot_success_rate(RUNS_DIR)
plt.show()
```

## Per-run drill-down
Pick a run from the index above; inspect its per-generation table and top solutions.

```python
run = idx["run"].iloc[0] if len(idx) else None
print("run:", run)
prog = load_progress(RUNS_DIR, runs=[run]) if run else None
prog
```

```python
sols = load_solutions(RUNS_DIR, run) if run else None
# top-10 solutions by raw_score; the .py files live in runs/<run>/solutions/<sol>.py
sols.sort_values("raw_score", ascending=False).head(10) if sols is not None and len(sols) else sols
```
